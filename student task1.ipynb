```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

sns.boxplot(x=df["absences"])

plt.title("Boxplot of Student Absences")
plt.xlabel("Number of Absences")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(data=df[["G1", "G2", "G3"]])

plt.title("Boxplot of Student Grades")
plt.xlabel("Grade")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="sex")

plt.title("Gender Distribution of Students")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x="age")

plt.title("Age Distribution of Students")
plt.xlabel("Age")
plt.ylabel("Number of Students")

plt.show()
```

```python
average_grades = df[["G1", "G2", "G3"]].mean()

print(average_grades)
```

```python
plt.figure(figsize=(8, 5))

average_grades.plot(kind="bar")

plt.title("Average Student Grades")
plt.xlabel("Grade")
plt.ylabel("Average Score")

plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="studytime", y="G3")

plt.title("Study Time vs Final Grade")
plt.xlabel("Study Time")
plt.ylabel("Average Final Grade (G3)")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(data=df, x="absences", y="G3")

plt.title("Absences vs Final Grade")
plt.xlabel("Number of Absences")
plt.ylabel("Final Grade (G3)")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(data=df, x="sex", y="G3")

plt.title("Gender vs Final Grade")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade (G3)")

plt.show()
```

```python
plt.figure(figsize=(10, 7))

correlation = df[
    ["age", "Medu", "Fedu", "studytime", "failures",
     "famrel", "freetime", "goout", "Dalc", "Walc",
     "health", "absences", "G1", "G2", "G3"]
].corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)

plt.title("Correlation Heatmap of Student Data")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

numeric_columns = [
    "age", "Medu", "Fedu", "traveltime",
    "studytime", "failures", "famrel",
    "freetime", "goout", "Dalc", "Walc",
    "health", "absences", "G1", "G2", "G3"
]

scaler = StandardScaler()

df_scaled = df.copy()

df_scaled[numeric_columns] = scaler.fit_transform(
    df_scaled[numeric_columns]
)

df_scaled.head()
```

```python
df.to_csv("cleaned_student_data.csv", index=False)

print("Cleaned student dataset saved successfully!")
```

```python
df_scaled.to_csv("scaled_student_data.csv", index=False)

print("Scaled student dataset saved successfully!")
```
